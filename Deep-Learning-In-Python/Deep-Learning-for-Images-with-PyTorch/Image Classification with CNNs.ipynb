#### Binary image classification: fully connected layer
- `Conv2d()`:
    - input: 3 RGB channels (red, green, blue)
    - Output: 16 channels
    - Kernel: 3 x 3 Matrix
    - Stride = 1: the kernel moves 1 step
    - padding = 1: 1pixel around the border
- `ReLU()`:
    - A non-linear actvation function

- `MaxPool2d()`:
    - Kernel: 2 x 2
    - Stride: 2 steps
- `Flatten()`:
    - Tensors flattened into 1-D vector

- `Linear()`:
    - input: feature maps x height x width
    - Output: a single class
- `SIgmoid()`:
    -[0, 1]


```python
class BinaryCNN(nn.Module):
    def __init__(self):
        super(BinaryCNN, self).__init__()
        self.conv1 = nn.Conv2d(3, 16, kernel_size=3, stride=1, padding =1)
        self.relu = nn.ReLU()
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(16 * 112 * 122, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.pool(self.relu(self.conv1(x)))
        x = self.fc1(self.flatten(x))
        x = self.sigmoid(x)
        
        return x
        
```

### Multi-class image classification with CNN

```python
class MultiClassCNN(nn.Module):
    def __init__(self, num_classes):
        super(MultiClassCNN, self).__init__()
        self.conv1 = nn.Conv2d(3, 16, kernel_size=3, stride=1, padding =1)
        self.relu = nn.ReLU()
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(16 * 112 * 122, num_classes)
        self.softmax = nn.softmax(dim=1)

    def forward(self, x):
        x = self.pool(self.relu(self.conv1(x)))
        x = self.fc1(self.flatten(x))
        x = self.softmax(x)

        return x
```

#### Kernel sizes
- The most common kernel size: 3x3(`Conv2d`) and 2x2(`MaxPool2d`)
- Convolution is a dot product of the kernel and the image region
- THe sum of the dot product creates a feature map 

#### Kernel is a filter
- Capture image patterns

#### Conv2d: output channels
- The number of out put channels determine how many filters are applied
- Each output channel corresponds to a distinct filter
- A higher number of output channels allows the layer to learn more complex features
- Output channel numbers are commonly chosen as powers of 2(6, 32, 63, 128)
    - It simplifies the process of combining and dividing channels in subsequent layers
