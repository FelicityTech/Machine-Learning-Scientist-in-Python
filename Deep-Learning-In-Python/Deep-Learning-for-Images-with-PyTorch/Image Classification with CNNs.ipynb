#### Binary image classification: fully connected layer
- `Conv2d()`:
    - input: 3 RGB channels (red, green, blue)
    - Output: 16 channels
    - Kernel: 3 x 3 Matrix
    - Stride = 1: the kernel moves 1 step
    - padding = 1: 1pixel around the border
- `ReLU()`:
    - A non-linear actvation function

- `MaxPool2d()`:
    - Kernel: 2 x 2
    - Stride: 2 steps
- `Flatten()`:
    - Tensors flattened into 1-D vector

- `Linear()`:
    - input: feature maps x height x width
    - Output: a single class
- `SIgmoid()`:
    -[0, 1]


```python
class BinaryCNN(nn.Module):
    def __init__(self):
        super(BinaryCNN, self).__init__()
        self.conv1 = nn.Conv2d(3, 16, kernel_size=3, stride=1, padding =1)
        self.relu = nn.ReLU()
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(16 * 112 * 122, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.pool(self.relu(self.conv1(x)))
        x = self.fc1(self.flatten(x))
        x = self.sigmoid(x)
        
        return x
        
```

### Multi-class image classification with CNN

```python
class MultiClassCNN(nn.Module):
    def __init__(self, num_classes):
        super(MultiClassCNN, self).__init__()
        self.conv1 = nn.Conv2d(3, 16, kernel_size=3, stride=1, padding =1)
        self.relu = nn.ReLU()
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(16 * 112 * 122, num_classes)
        self.softmax = nn.softmax(dim=1)

    def forward(self, x):
        x = self.pool(self.relu(self.conv1(x)))
        x = self.fc1(self.flatten(x))
        x = self.softmax(x)

        return x
```
