### Sentence tokenization
- Text -> sentences
- Offers clearer insights than analyzing the text as whole

```python
import nltk
nltk.download('punkt_tab')

text = "NLP is fun. Let's dice into it!"
sentences = nltk.sent_tokenize(text)
print(sentences)
```

### Word Tokenization
- Text -> words and punctuation
- Useful for tasks that require:
    - Identifying key terms
    - Counting word frequency

```python
text = "Claim your free prize now!"
words = nltk.word_tokenize(text)
print(words)
```

### Stop words and punctuation handling

Stop words
- Appear frequently but contribute little to machine's understanding of context
- Don't add much value in many NLP tasks
- Removing them helps models focus on important workds

Accessing stop words
- NLTK provides a list of stop words for several languages


```python
from nltk.corpus import stopwords
nltk.download('stopwords')

stop_words = stopwords.words('english')
print(stop_words[:10])
```

Removing stop words

```python
from nltk.tokenize import word_tokenize

text = "This is an example to demostrate removing stop words."
tokens = word_tokenize(text)
# THe .lower() method helps with case snesitivity
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]
print(filtered_tokens)
```

#### Punctuation
- Structuring language for humans

#### Punctuation removal
Useful for
- Task requiring to find common or important words in documents - Social media post

Not useful for
- Tasks requiring to maintain sentence structure for clarity - Text summarization


Accessing and removing punctuation

```python
import string
print(string.punctuation)
```

```python
text = "This is an example to demonstrate removing stop words."
tokens = word_tokenize(text)
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]

clean_tokens = [word for word in filtered_tokens if word not in string.punctuation]
print(clean_tokens)
```

Removing stop word


```python
feedback = "I reached out to support and got a helpful response within minutes!!! Very #impressed"

# Tokenize the text
tokens = word_tokenize(feedback)

# Get the list of English stop words
stop_words = stopwords.words('english')

# Remove stop words 
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]

print(filtered_tokens)
```
